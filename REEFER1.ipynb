```python
import pandas as pd
df = pd.read_excel('Reef.xlsx')
df
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt



plt.figure(figsize=(8,6))

sns.stripplot(
    data=df,
    x='Reefer_Truck',
    y='PreCooling_Time',
    hue='Reefer_Truck',
    jitter=0.15,
    size=10
)

plt.axhline(y=60, color='darkred', linestyle='--',
            label='60 min Reference')

plt.axhline(y=90, color='red', linestyle='--',
            label='90 min Buffer')
plt.legend()


plt.xlabel('Truck')
plt.ylabel('Pre-cooling Time (Minutes)')
plt.title('WK 44 Reefer Pre-cooling Time Performance')

plt.show()
```
